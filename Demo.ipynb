# Simple Statistical Tests in Python

### Comparing two groups: _is urban income statistically diﬀerent from rural income?_

```python
rural = df[df['region_type'] == 'Rural']['income']
urban = df[df['region_type'] == 'Urban']['income']

t, p = stats.ttest_ind(rural, urban, equal_var=False)

print("t =", round(t, 2))
print("p =", round(p, 4))
```

```python
temp = df[['region_type','income']].dropna()

rural = temp[temp['region_type'] == 'Rural']['income']
urban = temp[temp['region_type'] == 'Urban']['income']

t, p = stats.ttest_ind(rural, urban, equal_var=False)

print("t =", round(t, 2))
print("p =", round(p, 4))
```

### Comparing the Same Household Twice: _are `wellbeing_score`  significantly different before and after a program?_

```python
# Suppose we have before and after wellbeing scores
before = df['wellbeing_score'].dropna().iloc[:50]
after = before + 2 + (before * 0.01)

t, p = stats.ttest_rel(after, before)

print("t =", round(t, 2))
print("p =", round(p, 4))
```

# When Data Is Not Simple: Non-Parametric Tests

### Load the Data

```python
import pandas as pd
from scipy import stats

df = pd.read_csv("village_survey.csv")

print("Data loaded.")
```

### Mann–Whitney U test

```python
rural = df[df['region_type'] == 'Rural']['income'].dropna()
urban = df[df['region_type'] == 'Urban']['income'].dropna()

U, p = stats.mannwhitneyu(rural, urban, alternative='two-sided')

print("U =", round(U, 0))
print("p =", round(p, 4))
```

### Wilcoxon (Before and After)

```python
before = df['wellbeing_score'].dropna().iloc[:50]
after = before + 2 + (before * 0.01)

W, p = stats.wilcoxon(after, before)

print("W =", round(W, 0))
print("p =", round(p, 4))
```

### Kruskal–Wallis (Replaces ANOVA)

```python
groups = [
    df[df['region_type'] == 'Rural']['wellbeing_score'].dropna(),
    df[df['region_type'] == 'Urban']['wellbeing_score'].dropna()
]

H, p = stats.kruskal(*groups)

print("H =", round(H, 2))
print("p =", round(p, 4))
```

### Chi-Square (Two Categories)

```python
table = pd.crosstab(df['region_type'], df['wellbeing_score'])
print(table)

chi2, p, dof, expected = stats.chi2_contingency(table)

print("chi-square =", round(chi2, 2))
print("p =", round(p, 4))
```

```python
table = pd.crosstab(df['region_type'], df['education_years'])
# print(table)

chi2, p, dof, expected = stats.chi2_contingency(table)

print("chi-square =", round(chi2, 2))
print("p =", round(p, 4))
```

### Spearman (Ranked Correlation)

```python
rho, p = stats.spearmanr(df['education_years'].dropna(),
                         df['wellbeing_score'].dropna())

print("Spearman rho =", round(rho, 2))
print("p =", round(p, 4))
```

```python
import pandas as pd
from scipy import stats

df = pd.read_csv("village_survey.csv")

# Select the two columns, drop rows where either is missing
sub = df[['education_years', 'wellbeing_score']].dropna()

print("Rows used:", len(sub))

rho, p = stats.spearmanr(sub['education_years'], sub['wellbeing_score'])

print("Spearman rho =", round(rho, 2))
print("p =", round(p, 4))
```

# Making Charts: Turning Numbers Into Pictures

### Loading the tools

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("village_survey.csv")

print("Tools ready and data loaded.")
```

### A Simple Bar Chart

```python
sns.countplot(data=df, x='region_type')

plt.title("Number of households by region")
plt.xlabel("Region")
plt.ylabel("Count")
plt.show()
```

### A Histogram

```python
sns.histplot(data=df, x='income', bins=20)

plt.title("Distribution of household income")
plt.xlabel("Monthly income (INR)")
plt.ylabel("Number of households")
plt.show()
```

### A Box Plot

```python
sns.boxplot(data=df, x='region_type', y='income')

plt.title("Income by region type")
plt.xlabel("Region")
plt.ylabel("Monthly income (INR)")
plt.show()
```

### A Scatterplot

```python
sns.scatterplot(data=df, x='education_years', y='income')

plt.title("Income vs years of education")
plt.xlabel("Years of education")
plt.ylabel("Monthly income (INR)")
plt.show()
```

### Saving a Chart

```python
sns.boxplot(data=df, x='region_type', y='income')

plt.title("Income by region type")
plt.xlabel("Region")
plt.ylabel("Monthly income (INR)")

plt.savefig("income_by_region.png", dpi=300, bbox_inches='tight')
plt.show()
```

### Saving in Diﬀerent Formats

```python
sns.boxplot(data=df, x='region_type', y='income')

plt.title("Income by region type")
plt.xlabel("Region")
plt.ylabel("Monthly income (INR)")

plt.savefig("chart.png", dpi=300)   # for slides
plt.savefig("chart.pdf")            # for journals (vector)
plt.savefig("chart.svg")            # for websites
plt.show()
```

### Combining Two Charts

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.boxplot(data=df, x='region_type', y='income', ax=axes[0])
axes[0].set_title("Income by region")

sns.boxplot(data=df, x='gender', y='income', ax=axes[1])
axes[1].set_title("Income by gender")

plt.tight_layout()
plt.show()
```
